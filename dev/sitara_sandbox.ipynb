It is finally time to create a `get_leaders()` function for the above code. You will build on it later-on. This function takes no parameter. Inside it, you will need to:
1. define the urls
2. get the cookies
2. get the countries
3. loop over them and save their leaders in a dictionary
4. return the dictionary

```python
# < 15 lines
import requests

def get_leaders():
    # 1. define urls
    root_url = "https://country-leaders.onrender.com"
    cookie_url = root_url + "/cookie"
    countries_url = root_url + "/countries"
    leaders_url = root_url + "/leaders"

    # 2. get the cookies
    cookies = requests.get(cookie_url).cookies

    # 3. get the countries
    countries = requests.get(countries_url, cookies=cookies).json()

    # 4. loop over them and save their leaders in a dictionary
    leaders_per_country = {}

    for country in countries:
        r = requests.get(
            leaders_url,
            cookies=cookies,
            params={"country": country}
        )
        leaders_per_country[country] = r.json()

    # 5. return the dictionary
    return leaders_per_country
```

```python
# 10 lines

def clean_wikipedia_text(text):
    # Remove Wikipedia references/citations
    # Examples: [1], [12], [a], [citation needed]
    text = re.sub(r"\[[^\]]*\]", "", text)

    # Remove IPA / phonetic pronunciation blocks
    # Example: (/bəˈrɑːk oʊˈbɑːmə/)
    text = re.sub(r"\([^)]*[/ˈˌ][^)]*\)", "", text)

    # Replace common HTML entities
    text = re.sub(r"&nbsp;", " ", text)
    text = re.sub(r"&amp;", "&", text)

    # Replace multiple spaces, tabs and line breaks with a single space
    text = re.sub(r"\s+", " ", text)

    # Remove spaces before punctuation marks
    # Example: "hello ." -> "hello."
    text = re.sub(r"\s+([.,;:!?])", r"\1", text)

    return text.strip()
```

## 3. Putting it all together

Let's go back to your `get_leaders()` function and update it with an *inner* loop over each leader. You will query the url provided and extract the first paragraph using the `get_first_paragraph()` function you just finished. You will then update that `leader`'s dictionary and move on to the next one.

Notice, the rest of the code should not change since you modify the leader's data one by one.

```python
#  20 lines

def get_leaders():
    cookie_url, countries_url, leaders_url = "cookie", "countries", "leaders"

    def get_new_cookies():
        return requests.get(f"{root_url}/{cookie_url}").cookies

    cookies = get_new_cookies()
    countries = requests.get(f"{root_url}/{countries_url}", cookies=cookies).json()

    leaders_per_country = {}

    # Create shared session for all of requests
    with Session() as session:
        for country in countries:
            response = requests.get(
                f"{root_url}/{leaders_url}",
                cookies=cookies,
                params={"country": country}
            )

            if response.status_code != 200:
                cookies = get_new_cookies()
                response = requests.get(
                    f"{root_url}/{leaders_url}",
                    cookies=cookies,
                    params={"country": country}
                )

            leaders = response.json()

            for leader in leaders:
                leader["intro"] = get_first_paragraph(leader["wikipedia_url"], session)

            leaders_per_country[country] = leaders

    return leaders_per_country
```

Well done! It took a while however... Let's speed things up. The main *bottleneck* is the loop. We call on the Wikipedia website many times.

You will use the same *session* to call all the wikipedia pages. Check the *Advanced Usage* section of the Requests module's documentation.

Start by modifying the `get_first_paragraph()` function to accept a session parameter and adjust the `get()` method call.

```python
# < 20 lines
from bs4 import BeautifulSoup

def get_first_paragraph(url, session):
    html = session.get(url).text
    soup = BeautifulSoup(html, "html.parser")

    content = soup.find("div", id="mw-content-text")

    for p in content.find_all("p"):
        text = p.get_text(strip=True)
        if text:
            return text
```

Modify your `get_leaders()` function to make use of a single session for all the Wikipedia calls.
1. create a `Session` object outside of the loop over countries.
2. pass it to the `get_first_paragraph()` function as an argument.

```python
# <25 lines
import requests

def get_leaders():
    root_url = "https://country-leaders.onrender.com"
    countries_url = root_url + "/countries"
    cookie_url = root_url + "/cookie"
    leaders_url = root_url + "/leaders"

    # one session for everything (faster + persistent)
    session = requests.Session()

    def get_cookies():
        return requests.get(cookie_url).cookies

    cookies = get_cookies()
    countries = session.get(countries_url, cookies=cookies).json()

    leaders_per_country = {}

    for country in countries:

        req = session.get(leaders_url, cookies=cookies, params={"country": country})

        # refresh cookie if expired
        if req.status_code == 403:
            cookies = get_cookies()
            req = session.get(leaders_url, cookies=cookies, params={"country": country})

        if req.status_code == 200:
            leaders = req.json()

            #  Wikipedia calls ALSO use same session
            for leader in leaders:
                url = leader.get("wikipedia_url")
                if url:
                    leader["bio"] = get_first_paragraph(url, session)

            leaders_per_country[country] = leaders

    return leaders_per_country
```
